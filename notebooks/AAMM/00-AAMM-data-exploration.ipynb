### Puntos con topónimos - archivos con sufijo "*_to"
Puede ser útil como una capa complementaria de contexto geográfico, porque contiene puntos con topónimos que permiten identificar y caracterizar elementos del territorio que pueden estar relacionados con los reportes de inundación.

Un topónimo es simplemente el nombre propio que recibe un lugar o un elemento geográfico.

Por ejemplo:

Hermosillo → topónimo de una ciudad.
Río Sonora → topónimo de un río.
Cerro de la Campana → topónimo de un accidente geográfico.
Laguna X → topónimo de una laguna.
Canal Y → topónimo de un canal.

| Columna        | Justificación                                                                                                                                                |
| -------------- | ------------------------------------------------------------------------------------------------------------------------------------------------------------ |
| **geometry**   | Permite ubicar espacialmente los topónimos y relacionarlos mediante análisis espacial con los reportes de inundación y los elementos de la red hidrográfica. |
| **FC**         | Permite identificar la clasificación del elemento geográfico.                                                                                                |
| **CLASE**      | Permite distinguir la clase a la que pertenece el elemento geográfico.                                                                                       |
| **TERMINO_GE** | Permite identificar el tipo de rasgo geográfico, por ejemplo, canal, arroyo, bordo, laguna o zona sujeta a inundación.                                       |
| **NOMBRE**     | Permite identificar el nombre oficial del elemento geográfico cuando sea necesario interpretar o contextualizar un elemento identificado espacialmente.      |

## Llamadas 911 por colonia (HuggingFace)

Fuente: `Marcelinux/llamadas911_colonias_hermosillo_2018_2025` — un solo archivo Excel (`213.xlsx`, ~52 MB), licencia CC-BY-4.0. Por el nombre del dataset, son registros de llamadas al 911 en colonias de Hermosillo entre 2018 y 2025.

Se descarga a `notebook_experimentation_data` y se explora: estructura, tipos de datos, nulos y un borrador del significado de cada columna.

```python
import sys

import requests

# openpyxl no es dependencia del proyecto; se instala como en la celda 1
!{sys.executable} -m pip install openpyxl
```

```python
llamadas_URL = (
    "https://huggingface.co/datasets/"
    "Marcelinux/llamadas911_colonias_hermosillo_2018_2025/resolve/main/213.xlsx"
)
xlsx_path = notebook_data_dir / "llamadas911_213.xlsx"

if not xlsx_path.exists():
    r = requests.get(llamadas_URL, timeout=300)
    r.raise_for_status()
    xlsx_path.write_bytes(r.content)

llamadas_xl = pd.ExcelFile(xlsx_path)
print("Hojas:", llamadas_xl.sheet_names)
```

```python
# una hoja por anio (2018-2025), mismas columnas: se concatenan
llamadas_df = pd.concat(
    [llamadas_xl.parse(hoja) for hoja in llamadas_xl.sheet_names], ignore_index=True
)
print(f"{llamadas_df.shape[0]} filas x {llamadas_df.shape[1]} columnas")
```

```python
# Revision inicial: columnas, tipos y nulos
print(list(llamadas_df.columns))

llamadas_df.info()
nulos = llamadas_df.isna().sum()
print(pd.DataFrame({"nulos": nulos, "pct": (nulos / len(llamadas_df) * 100).round(1)}
    )
)
llamadas_df.head()
```

```python
# Valores por columna para inferir significados
print(llamadas_df.describe(include="all"))
print()
for col in ["COLONIA", "TIPO DE INCIDENTE"]:
    print(f"--- {col} ({llamadas_df[col].nunique()} distintos)")
    print(llamadas_df[col].value_counts().head(15))
    print()
```

### Borrador: significado aparente de cada columna

**2,297,081 filas × 4 columnas** — una fila por llamada/reporte de incidente, sin identificador propio. El Excel trae una hoja por año (2018–2025) con las mismas columnas.

| columna | tipo | qué parece ser |
|---|---|---|
| `COLONIA` | str | Colonia donde se reportó el incidente, en mayúsculas (2,296 valores distintos; el más frecuente es `CENTRO` con ~50.7k). Parece texto libre del operador del 911, incluye valores genéricos como `SIN COLONIA` (36.4k).
| `TIPO DE INCIDENTE` | str | Clasificación del motivo de la llamada (475 valores distintos; el más común es `RUIDO EXCESIVO` ~294.8k). Mezcla incidentes de seguridad (`ALLANAMIENTO DE MORADA`, `PORTACIÓN DE ARMAS`) con apoyo ciudadano (`PERSONA AGRESIVA`, `VEHÍCULO A EXCESO DE VELOCIDAD`).

```python
elegidos = {t for tipos in seleccion_manual.values() for t in tipos}
no_encontrados = elegidos - set(tipos_incidente)
print(f"Tipos seleccionados: {len(elegidos)}")
print()
llamadas_vc = llamadas_df["TIPO DE INCIDENTE"].value_counts()
total = 0
for grupo, tipos in seleccion_manual.items():
    subtotal = llamadas_vc.reindex(tipos).sum()
    total += subtotal
    print(f"--- {grupo}: {int(subtotal):,} llamadas")
    print(llamadas_vc.reindex(tipos).sort_values(ascending=False).to_string())
    print()
print(f"TOTAL seleccionados: {int(total):,} llamadas ({total/len(llamadas_df):.2%} del total)")
```
