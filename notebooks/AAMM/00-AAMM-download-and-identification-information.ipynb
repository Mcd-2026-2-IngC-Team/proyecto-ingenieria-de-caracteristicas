### Extraer información de .zip

```python
for zip_f in notebook_data_dir.glob("*.zip"):
    output_dir = notebook_data_dir / zip_file.stem
    
    with zipfile.ZipFile(zip_f, "r") as zip_ref:
        zip_ref.extractall(output_dir)
```

```python
for zip_file in notebook_data_dir.glob("*.zip"):
    print(f"Descomprimiendo: {zip_file.name}")

    output_dir = notebook_data_dir / zip_file.stem
    output_dir.mkdir(parents=True, exist_ok=True)

    with zipfile.ZipFile(zip_file, "r") as zip_ref:
        zip_ref.extractall(output_dir)

    print(f"Listo: {output_dir}")
```
